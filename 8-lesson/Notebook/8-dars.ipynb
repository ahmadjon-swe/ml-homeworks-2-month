# 8. Hosilaning qo'llanilishi (Cost function)

```python
import numpy as np
import pandas as pd
```

```python
# Haqiqiy qiymatlar
y_true = np.array([10, 15, 20, 25])

# Model bashorati
y_pred = np.array([12, 14, 19, 26])

print("Haqiqiy qiymatlar", y_true)
print("Bashorat", y_pred)
```

```python
mse = np.mean((y_true - y_pred) ** 2)
print("Mean Square Error =" , mse)
```

```python
import sympy as sp
w = sp.Symbol("w")
j = w**2
gradient = sp.diff(j, w)
print("Cost function =", j)
print(" =", gradient)
```

```python
import matplotlib.pyplot as plt 

w = np.linspace(-5,5,3000)
j=w**2
plt.figure(figsize=(8,5))
plt.plot(w,j,linewidth=3)
plt.scatter(0,0,color="red",s=80)
plt.title("cost function")
plt.xlabel('weight')
plt.ylabel('cost')
plt.grid(True)
plt.show()
```

```python
w = 8
learning_rate = 0.1
for epoch in range(10):
    gradient = 2*w
    w = w - learning_rate * gradient
    cost = w **2
    print(f"Epoch {epoch+1:2d} weight={w:.4f} cost = {cost:.4f}")
```

```python
cost_history = []
w = 8
lr = 0.1
for _ in range(20):
    gradient = 2*w
    w = w - lr*gradient
    cost_history.append(w**2)


plt.figure(figsize=(8,5))
plt.plot(cost_history, marker="o")
plt.title("cost function kamayishi")
plt.xlabel('iteration')
plt.ylabel('cost')
plt.grid(True)
plt.show()
```

Machine learning'da sympy

```python
w, b, x, y = sp.symbols("w b x y")
z=w*x+b
sig = 1/(1+sp.exp(-z))
loss = (y - sig)**2
fradient = sp.diff(loss, w)
print(fradient)
```

Oddiy limit

```python
expr = 2*x +1
limit = sp.limit(expr, x, 2)
print(limit)
```

```python
expr = (x**2 - 4) / (x-2)
left_limit = sp.limit(expr, x, 2, dir="-")
print(left_limit)
```

o'ng limit

```python
expr = (x**2 - 4) / (x-2)
right_limit = sp.limit(expr, x, 2, dir="+")
print(right_limit)
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-1, 5, 100)
y = 2 * x + 1

plt.figure(figsize=(8,5))
plt.plot(x, y)

plt.scatter(2,5, color="red")

plt.title("f(x)=2x+1")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid()

plt.show()
```

```python
import numpy as np
left = np.array([1.9, 1.99, 1.999])
right = np.array([2.1, 2.01, 2.001])
print("Chap limit")

for i in left:
    print(i, "-y", 2 * i + 1)


print()
print("O'ng limit:")

for i in right:
    print(i, "-y", 2 * i +1)
```

Loss function

```python
w = np.linspace(-5, 5, 200)
loss = (w-2)**2


plt.figure(figsize=(8,5))
plt.plot(w, loss)

plt.scatter(2,0, color="red")

plt.title("loss function")
plt.xlabel("weight")
plt.ylabel("loss")
plt.grid()

plt.show()
```

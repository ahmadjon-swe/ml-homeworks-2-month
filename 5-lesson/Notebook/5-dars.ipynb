```python
import numpy as np
```

```python
v = np.array([2, 3])

print(v)
```

```python
a = np.array([
    [2,0],
    [0,3]
])

print(a @ v)
```

Rotation

```python
theta = np.radians(90)

R = np.array([
    [np.cos(theta), -np.sin(theta)],
    [np.sin(theta), np.cos(theta)]
])

print(R @ v)
```

Reflection

```python
reflection =np.array([
    [-1, 0],
    [0, 1]
])

v = np.array([2, 3])
print(reflection @ v)
```

Original vector

```python
import numpy as np
import matplotlib.pyplot as plt


plt.figure(figsize=(6,6))

plt.quiver(
    0,0,
    v[0],v[1],
    angles='xy',
    scale_units='xy',
    scale=1,
    color='blue',
    label='Original'
)

plt.xlim(-6,6)
plt.ylim(-6,6)

plt.grid()

plt.legend()

plt.title("Scaling Transformation")

plt.show()
```

Scaling Visualization

```python
import numpy as np
import matplotlib.pyplot as plt


v = np.array([2, 3])

a = np.array([
    [2, 0],
    [0, 2]
])

new_v = a @ v

plt.figure(figsize=(6,6))

plt.quiver(
    0,0,
    v[0],v[1],
    angles='xy',
    scale_units='xy',
    scale=1,
    color='blue',
    label='Original'
)

plt.quiver(
    0,0,
    new_v[0],
    new_v[1],
    angles='xy',
    scale_units='xy',
    scale=1,
    color='red',
    label='Scaled'
)

plt.xlim(-8,8)
plt.ylim(-8,8)

plt.grid()

plt.legend()

plt.title("Scaling Transformation")

plt.show()
```

Rotation

```python
import numpy as np
import matplotlib.pyplot as plt

theta = np.radians(45)

R = np.array([
    [np.cos(theta), -np.sin(theta)],
    [np.sin(theta), np.cos(theta)]
])

v = np.array([3, 2])


new_v = R @ v

plt.figure(figsize=(6,6))

plt.quiver(
    0,0,
    v[0],v[1],
    angles='xy',
    scale_units='xy',
    scale=1,
    color='blue',
    label='Original'
)

plt.quiver(
    0,0,
    new_v[0],
    new_v[1],
    angles='xy',
    scale_units='xy',
    scale=1,
    color='green',
    label='Rotated'
)

plt.xlim(-8,8)
plt.ylim(-8,8)

plt.grid()

plt.legend()

plt.title("Rotating Transformation")

plt.show()
```

```python
v=np.array([2,3])
scale=np.array([[2,0],[0,2]])
theta=np.radians(45)
rotation=np.array([
    [np.cos(theta),-np.sin(theta)],
    [np.sin(theta), np.cos(theta)]
])
reflection=np.array([
    [-1,0],
    [0,1]
])
shear=np.array([
    [1,1],
    [0,1]
])
vectors={
    "Original":v,
    "Scaling":scale@v,
    "Rotation":rotation@v,
    "Reflection":reflection@v,
    "Shearing":shear@v
}

colors=["blue","red","green","orange","purple"]

plt.figure(figsize=(8,8))

for (name,vec),color in zip(vectors.items(),colors):

    plt.quiver(
        0,
        0,
        vec[0],
        vec[1],
        angles="xy",
        scale_units="xy",
        scale=1,
        color=color,
        label=name
    )

plt.grid()

plt.xlim(-8,8)
plt.ylim(-8,8)

plt.legend()

plt.title("Vector Transformations")

plt.show()
```

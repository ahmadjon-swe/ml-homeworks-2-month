# Vektor transformatsiyalari va tasvirlar

NumPy va Matplotlib yordamida 10 ta misol.

## 1. Scaling

```python
import numpy as np
import matplotlib.pyplot as plt
v=np.array([2,3]);S=np.array([[2,0],[0,2]]);v2=S@v
plt.quiver(0,0,*v,angles='xy',scale_units='xy',scale=1,label='Original')
plt.quiver(0,0,*v2,angles='xy',scale_units='xy',scale=1,label='Scaled')
plt.grid()
plt.axis('equal')
plt.legend()
plt.show()
```

## 2. Rotation 45

```python
import numpy as np
import matplotlib.pyplot as plt
v=np.array([4,2]);t=np.radians(45);R=np.array([[np.cos(t),-np.sin(t)],[np.sin(t),np.cos(t)]]);v2=R@v
plt.quiver(0,0,*v,angles='xy',scale_units='xy',scale=1)
plt.quiver(0,0,*v2,angles='xy',scale_units='xy',scale=1)
plt.grid();plt.axis('equal');plt.show()
```

## 9. Scale+Rotate

```python
import numpy as np
import matplotlib.pyplot as plt
v=np.array([2,2]);S=np.array([[2,0],[0,2]]);t=np.radians(30);R=np.array([[np.cos(t),-np.sin(t)],[np.sin(t),np.cos(t)]]);v2=R@(S@v)
plt.quiver(0,0,*v);plt.quiver(0,0,*v2);plt.grid();plt.axis('equal');plt.show()
```

## 10. All transforms

```python
import numpy as np
import matplotlib.pyplot as plt
v=np.array([2,3]);scale=np.array([[2,0],[0,2]]);theta=np.radians(45);rotation=np.array([[np.cos(theta),-np.sin(theta)],[np.sin(theta),np.cos(theta)]]);reflection=np.array([[-1,0],[0,1]]);shear=np.array([[1,1],[0,1]])
vectors={'Original':v,'Scaling':scale@v,'Rotation':rotation@v,'Reflection':reflection@v,'Shearing':shear@v}
for n,vec in vectors.items(): plt.quiver(0,0,vec[0],vec[1],angles='xy',scale_units='xy',scale=1,label=n)
plt.grid();plt.axis('equal');plt.legend();plt.show()
```

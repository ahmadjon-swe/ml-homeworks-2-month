```python
from sklearn.preprocessing import StandardScaler
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_wine
```

```python
wine = load_wine()
df = pd.DataFrame(
    wine.data, 
    columns=wine.feature_names
)

df.head()
```

```python
df[['alcohol', 'proline']].head()
```

```python
scaler = StandardScaler()

scaled = scaler.fit_transform(df)
scaled_df = pd.DataFrame(scaled, columns=df.columns)

scaled_df.head()
```

```python
scaled_df.mean()
```

```python
scaled_df.std()
```

```python
fig,ax=plt.subplots(figsize=(12,5))
ax.hist(df["proline"], bins=30)
ax.set_title("Original roline")
plt.show()
```

```python
fig,ax=plt.subplots(figsize=(12,5))
ax.hist(scaled_df["proline"], bins=30)
ax.set_title("Scaled roline")
plt.show()
```

```python
comparison = pd.DataFrame({
    "Original": df["proline"],
    "scaled": scaled_df["proline"]
})

comparison.head(10)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp

plt.style.use("ggplot")

print("Kutubxonalar yuklandi")
```

```python
x = sp.Symbol("x")
f = x**2
df = sp.diff(f, x)

print("f(x) =", f)
print("f(x) =", df)
```

```python
x = np.linspace(-4, 4, 300)
y = x**2

plt.figure(figsize=(8,5))
plt.plot(x,y,linewidth=3)
plt.scatter(2,4,color="red",s=80,label="Positive Slope")
plt.scatter(-2,4,color="blue",s=80,label="Negative Slope")
plt.legend()
plt.xlabel('x')
plt.ylabel('x')
plt.title("y = x ** 2")
plt.show()
```

```python
x = sp.Symbol("x")
h = sp.Symbol("h")

f = x**2
formula = ((x+h)**2 - x**2)
print("Farqlar nisbati")
print(sp.expand(formula))
print()
print("Hosila")
print(sp.limit(formula, h, 0))
```

```python
x = np.linspace(-1, 5, 300)
y = x**2
x0 = 2
y0 = x0**2
slope = 2*x0
tangent = slope*(x-x0)+y0
plt.figure(figsize=(8,5))
plt.plot(x,y,label="y=x**2")
plt.plot(x,tangent,label="Tangent", linestyle="--")
plt.scatter(x0,y0,s=80)
plt.legend()
plt.grid(True)
plt.show()
```

```python
x,y = sp.symbols("x y")
f=x**2+y**2
gradient = sp.Matrix([
    sp.diff(f,x),
    sp.diff(f,y)
])

print("Gradient")
gradient
```

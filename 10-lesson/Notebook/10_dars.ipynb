```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
```

```python
x, y = sp.symbols("x y")

f = x**2 + y**2

print("Funktsiya")
display(f)
```

```python
x, y = sp.symbols("x y")

f = 3*x**2 + 2*y

display(f)
```

```python
x, y = sp.symbols("x y")

f = x**2 + y**2

df_dx = sp.diff(f, x)
df_dy = sp.diff(f, y)

print("df/dx =")
display(df_dx)

print("df/dy =")
display(df_dy)
```

```python
from mpl_toolkits.mplot3d import Axes3D

x = np.linspace(-3,3,100)
y = np.linspace(-3,3,100)


X, Y = np.meshgrid(x,y)

Z = X**2 + Y**2
fig = plt.figure(figsize=(8,6))
ax=fig.add_subplot(111,projection="3d")
ax.plot_surface(X,Y,Z,cmap="viridis")
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")
plt.show()
```
